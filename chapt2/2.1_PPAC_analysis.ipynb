
<h1 id="2.1-PPAC-%E4%BF%A1%E5%8F%B7%E5%A4%84%E7%90%86-I">2.1 PPAC 信号处理-I</h1>



<h3 id="1.-PPAC">1. PPAC</h3><ul>
<li><strong>电离与雪崩</strong>：带电粒子穿过 PPAC 的工作气体（通常为 $C_3F_8$ 或 $C_4H_{10}$），在路径上产生初级电离。电子在强电场作用下向阳极漂移并发生 <strong>Townsend 雪崩</strong>。</li>
<li><strong>感应信号</strong>：<ul>
<li><strong>阳极 (Anode)</strong>：一整块薄镀铝 Mylar 膜。雪崩电子被阳极收集，产生快速的时间信号 $t_a$，作为该粒子到达 PPAC 的参考时间。</li>
<li><strong>阴极 (Cathode)</strong>：分成 1mm 间隔的读数条 (Strips)，X 和 Y 方向正交。阳极上的雪崩会在阴极条上产生感应电荷信号。</li>
</ul>
</li>
</ul>
<p><strong>Delay-line 读出原理</strong></p>
<p>阴极条之间通过高精度的电磁延迟线（Delay Line）连接。感应信号会沿延迟线向两端传输。</p>
<ul>
<li><strong>$T_{xdelay}/T_{ydelay}$</strong>：信号在x/y方向延迟线上的总传播时间（常数）。</li>
<li><strong>$Q$</strong>：感应信号的幅度。</li>
</ul>
<p>假设粒子击中位置为 $x$ (以中心为0)，延迟线总长度为 $L$，信号在延迟线内部传输时间（设信号传输速度 $v_d = L/T_{xdelay}$）：
$$
\begin{align}
&amp;T_{xdelay} = t_{x1} + t_{x2} \\
&amp;t_{x1} = \left(\frac{L}{2} + x \right) \frac{T_{xdelay}}{L} \\
&amp;t_{x2} = \left(\frac{L}{2} - x \right) \frac{T_{xdelay}}{L} \\
&amp;x = (t_{x1} - t_{x2}) \frac{L}{2 T_{xdelay}}
\end{align}
$$</p>

<h2 id="PPAC-Position-Calibration">PPAC Position Calibration</h2><ul>
<li>将 $\alpha$-放射源置于远处，$\alpha$ 粒子入射到盖有 Mask 的探测器上（Mask上孔的位置分布已知），得到 x 方向和 y 方向的一系列峰位。通过拟合时间差谱 $(\Delta t_{x1} - \Delta t_{x2})$ 的峰位和实际物理位置，得到线性刻度 $x=k\,\Delta t+b$。理想模型给出 $k=L/(2T_{xdelay})$；实际采用 mask 峰位的拟合结果，并检查各孔位置的 residual。</li>
<li>探测器安装到束流线后，通过位置测量，确定探测器中心相对于束流中心线的偏移 $b$。</li>
<li>实验中，通过测量经过中心线的束流（利用限制束流线上的 slit 狭缝），进一步检验和校准上述偏移。</li>
</ul>

<p><strong>实际PPAC 信号</strong></p>
<ul>
<li><strong>Trigger (time start)</strong>: Plastic scintillator (塑闪)</li>
<li><strong>飞行时间 ($TOF$)</strong>：粒子从塑闪飞到 PPAC 的时间</li>
<li><strong>阴极，阳极信号的 offset</strong>：$t_{off,x1}，t_{off,x2}，t_{off,a}$</li>
<li><strong>TDC 记录的绝对时间信号</strong>：$T_{x1}，T_{x2}，T_a$</li>
</ul>
<p><strong>阴极时间 (包含束流速度依赖性)</strong></p>
<p>TDC 记录的时间不仅包含了信号在探测器内的传输时间，还包含了粒子的飞行时间 ($TOF$) 和硬件 offset：</p>
<ul>
<li>$T_{x1} = TOF + t_{x1} + t_{off,x1}$</li>
<li>$T_{x2} = TOF + t_{x2} + t_{off,x2}$</li>
<li>$T_{x1} + T_{x2} = 2 TOF + T_{xdelay} + (t_{off,x1} + t_{off,x2})$</li>
<li>$T_{x1} - T_{x2} = t_{x1} - t_{x2} + (t_{off,x1} - t_{off,x2})$</li>
</ul>
<p><strong>阳极时间</strong></p>
<ul>
<li>$T_a = TOF + t_{off,a}$</li>
</ul>
<p><strong>组合参数 (相对时间，消除速度依赖性)</strong></p>
<p>上述绝对时间信号不仅与探测器内部传输有关，也与束流的到达时间 ($TOF$) 有关。研究探测器特征时，可先消除公共 $TOF$ 引入的展宽。
通过将阴极信号减去阳极信号，可以抵消 $TOF$的影响：</p>
<ul>
<li>$\Delta T_{x1} = T_{x1} - T_a = t_{x1} + (t_{off,x1} - t_{off,a})$</li>
<li>$\Delta T_{x2} = T_{x2} - T_a = t_{x2} + (t_{off,x2} - t_{off,a})$</li>
<li>$\Delta T_{x1} - \Delta T_{x2} = t_{x1} - t_{x2} + (t_{off,x1} - t_{off,x2})$</li>
</ul>
<h3 id="%E6%8E%A2%E6%B5%8B%E5%99%A8%E4%B8%A4%E7%AB%AF%E4%BF%A1%E5%8F%B7%E5%85%B3%E8%81%94">探测器两端信号关联</h3><ul>
<li>$\mathbf{T_{x1} + T_{x2}} = 2 TOF + Const.$
<strong>(包含 $TOF$，与束流速度有关，分布表现为一个较宽的峰)</strong></li>
<li>$\mathbf{\Delta T_{x1} + \Delta T_{x2}} = t_{x1} + t_{x2} + (t_{off,x1} + t_{off,x2} - 2t_{off,a}) = T_{xelay} + Const. = \mathbf{Const.}$
<strong>（公共到达时间被抵消，分布通常变窄；宽度包含两端及 anode 定时误差和相关项。）</strong></li>
</ul>

<h3>Time-sum 选择</h3>
<p>正常的两端信号应满足近似固定的 time-sum。若两粒子在延迟线传输时间内到达，两端可能记录了不同粒子的最早信号，time-sum 就可能偏小。δ rays 也可造成提前的读出信号。因此，峰外事件不能全称为 pileup，落在峰内也不能保证没有重叠。</p>
<p>下面以主峰的 ±3σ 作为示例 gate。先看上图主峰和尾部，再选择保留范围；σ 来自主峰附近的 Gaussian 拟合，不是对整个尾部的拟合。gate 的宽度决定了选择效率。</p>
<p>参考：H. Kumagai et al., <a href="https://arxiv.org/abs/1311.0215">Development of Parallel Plate Avalanche Counter PPAC for BigRIPS fragment separator</a>, arXiv:1311.0215 (2013)。文中讨论了 delay-line time-sum 与 δ-ray 效应。</p>

```c++
TString scut_xsum = Form("abs(dt_x2 + dt_x1 - %f) < 3 * %f", xpeak, xsigma);
TString scut_ysum = Form("abs(dt_y2 + dt_y1 - %f) < 3 * %f", ypeak, ysigma);
```

```c++
TCut cut_xsum = scut_xsum.Data() && cut_xvalid; 
TCut cut_ysum = scut_ysum.Data() && cut_yvalid; 
TCut cut_xysum = cut_xsum && cut_ysum;
```

<h2>3. 位置刻度与边界 cut</h2>
<p>这里用 $L$ 表示<strong>半宽</strong>，物理灵敏区为 $-L&lt;x&lt;L$（前面 delay-line 公式中的总宽度为 $2L$）。有限分辨使测量位置 $x_{\rm rec}=x+\delta x$ 在物理边界外也有分布。不能把谱中最外侧的非零 bin 强行刻度到 ±L；mask 孔或 slit 的已知位置才提供独立的位置基准。只有照明分布与边缘响应已知时，才适合用边缘模型辅助刻度。</p>
<p>下图仅用均匀照明、Gaussian 分辨 σ=1 mm、半宽 L=120 mm <strong>说明边界展宽</strong>，不是对本组 PPAC 数据的拟合。左图为测量分布在右边缘的变化；右图为真实位置给定时，通过 $|x_{\rm rec}|&lt;L$ 的概率。</p>

```c++
TCanvas *cBoundary=new TCanvas("cBoundary","Resolution at an active edge",1000,380);
cBoundary->Divide(2,1);
TF1 *edge=new TF1("edge",
 "0.5*(TMath::Erf((x+[0])/(sqrt(2)*[1]))-TMath::Erf((x-[0])/(sqrt(2)*[1])))",
 114,126);
edge->SetParameters(120,1); // half-width L, assumed sigma, in mm
cBoundary->cd(1);
edge->SetTitle("Uniform illumination, Gaussian response;x_{rec} (mm);Relative density");
edge->Draw();
TLine *boundary=new TLine(120,0,120,1);
boundary->SetLineStyle(2); boundary->Draw();
cBoundary->cd(2);
TF1 *survival=(TF1*)edge->Clone("survival");
survival->SetRange(114,120);
survival->SetTitle("Acceptance of |x_{rec}|<120 mm;True x (mm);Cut survival probability");
survival->Draw();
cout << "At x=L: " << survival->Eval(120)
     << "; at x=L-sigma: " << survival->Eval(119) << endl;
cBoundary->Draw();
```

<h3>怎样选择边界？</h3>
<p>若目的是测量探测器内部的读出效率，先用<strong>不包含待测 PPAC 的 reference track</strong>预测位置，并选择 $|x_{\rm ref}|&lt;L-d$ 的内部区域。$d$ 根据参考径迹的不确定度及边缘响应选择，不能直接用待测层自己的位置定义分母。下一节采用的 $|x_{\rm ref}|&lt;100$ mm、$|y_{\rm ref}|&lt;60$ mm 就是在 240×150 mm² 灵敏区内留下边缘余量。</p>
<p>若分析要求测量坐标也落在 ±L 内，可以再加这个 cut，但删去的边缘事件属于<strong>位置选择损失</strong>，不应直接解释为探测器没有响应。也可以保留边界外的少量测量尾部，再通过 residual 和本底确定是否可靠。没有一条对所有束斑和分辨都适用的固定放宽范围。</p>
<p>在同一 reference 样本内，分别数出 $N_{\rm ref}$、有有效读数的 $N_{\rm valid}$、以及读数通过边界 cut 的 $N_{\rm pass}$：</p>
<p>$$\epsilon_{\rm read}=\frac{N_{\rm valid}}{N_{\rm ref}},\qquad
\epsilon_{\rm cut}=\frac{N_{\rm pass}}{N_{\rm valid}},\qquad
\epsilon_{\rm selected}=\frac{N_{\rm pass}}{N_{\rm ref}}
=\epsilon_{\rm read}\epsilon_{\rm cut}.$$</p>
<p>这是计数定义，不要求两个过程独立。若已用最后一个比值作总效率修正，就不要再重复修正边界损失。真实边界处的 cut 保留概率约为 50%，不表示全探测器损失一半；总损失还取决于束斑在边缘的粒子比例。</p>
