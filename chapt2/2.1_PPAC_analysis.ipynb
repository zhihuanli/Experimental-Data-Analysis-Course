
<h1 id="2.1-PPAC-%E4%BF%A1%E5%8F%B7%E5%A4%84%E7%90%86-I">2.1 PPAC 信号处理-I</h1>



<h3 id="1.-PPAC">1. PPAC</h3><ul>
<li><strong>电离与雪崩</strong>：带电粒子穿过 PPAC 的工作气体（通常为 $C_3F_8$ 或 $C_4H_{10}$），在路径上产生初级电离。电子在强电场作用下向阳极漂移并发生 <strong>Townsend 雪崩</strong>。</li>
<li><strong>感应信号</strong>：<ul>
<li><strong>阳极 (Anode)</strong>：一整块薄镀铝 Mylar 膜。雪崩电子被阳极收集，产生快速的时间信号 $t_a$，作为该粒子到达 PPAC 的参考时间。</li>
<li><strong>阴极 (Cathode)</strong>：分成 1mm 间隔的读数条 (Strips)，X 和 Y 方向正交。阳极上的雪崩会在阴极条上产生感应电荷信号。</li>
</ul>
</li>
</ul>
<p><strong>Delay-line 读出原理</strong></p>
<p>阴极条之间通过高精度的电磁延迟线（Delay Line）连接。感应信号会沿延迟线向两端传输。</p>
<ul>
<li><strong>$T_{xdelay}/T_{ydelay}$</strong>：信号在x/y方向延迟线上的总传播时间（常数）。</li>
<li><strong>$Q$</strong>：感应信号的幅度。</li>
</ul>
<p>假设粒子击中位置为 $x$ (以中心为0)，延迟线总长度为 $L$，信号在延迟线内部传输时间（设信号传输速度 $v_d = L/T_{xdelay}$）：
$$
\begin{align}
&amp;T_{xdelay} = t_{x1} + t_{x2} \\
&amp;t_{x1} = \left(\frac{L}{2} + x \right) \frac{T_{xdelay}}{L} \\
&amp;t_{x2} = \left(\frac{L}{2} - x \right) \frac{T_{xdelay}}{L} \\
&amp;x = (t_{x1} - t_{x2}) \frac{L}{2 T_{xdelay}}
\end{align}
$$</p>

<h2 id="PPAC-Position-Calibration">PPAC Position Calibration</h2><ul>
<li>将 $\alpha$-放射源置于远处，$\alpha$ 粒子入射到盖有 Mask 的探测器上（Mask上孔的位置分布已知），得到 x 方向和 y 方向的一系列峰位。通过拟合时间差谱 $(\Delta t_{x2} - \Delta t_{x1})$ 的峰位和实际物理位置，得到刻度系数 $k = L / 2T_{xdelay}$。</li>
<li>探测器安装到束流线后，通过位置测量，确定探测器中心相对于束流中心线的偏移 $b$。</li>
<li>实验中，通过测量经过中心线的束流（利用限制束流线上的 slit 狭缝），进一步检验和校准上述偏移。</li>
</ul>

<h3 id="2.-%E5%AE%9E%E9%AA%8C%E8%AE%BE%E7%BD%AE:">2. 实验设置:</h3><p>实验中，<strong>F8PPAC1-3</strong> 用于测量靶前束流的入射方向和轨迹。靶后有两套探测系统：</p>
<ol>
<li><strong>MUST2 望远镜系统</strong>：测量反应产生的轻粒子。</li>
<li><strong>零度磁谱仪 (ZDS)</strong>：探测剩余重核。</li>
</ol>

为了减少散射粒子在靶内的能量损失，靶框被旋转 <strong>45°</strong> 朝向望远镜摆放。靶前放置了由厚铁块组成的屏蔽体，用于阻挡束流晕（Beam Halo），避免偏离中心线的非物理粒子击中靶后的探测系统。</p>

<ul>
<li>一种是 double PPAC (F8PPAC1, F8PPAC2)，面积为 $240 \times 150 \text{ mm}^2$。</li>
<li>另一种是 single PPAC (F8PPAC3)，面积为 $100 \times 100 \text{ mm}^2$。
<img alt="setup" src="fig/ppac_types.png"/></li>
</ul>
<p><strong>实际PPAC 信号</strong></p>
<ul>
<li><strong>Trigger (time start)</strong>: Plastic scintillator (塑闪)</li>
<li><strong>飞行时间 ($TOF$)</strong>：粒子从塑闪飞到 PPAC 的时间</li>
<li><strong>阴极，阳极信号的 offset</strong>：$t_{off,x1}，t_{off,x2}，t_{off,a}$</li>
<li><strong>TDC 记录的绝对时间信号</strong>：$T_{x1}，T_{x2}，T_a$</li>
</ul>
<p><strong>阴极时间 (包含束流速度依赖性)</strong></p>
<p>TDC 记录的时间不仅包含了信号在探测器内的传输时间，还包含了粒子的飞行时间 ($TOF$) 和硬件 offset：</p>
<ul>
<li>$T_{x1} = TOF + t_{x1} + t_{off,x1}$</li>
<li>$T_{x2} = TOF + t_{x2} + t_{off,x2}$</li>
<li>$T_{x1} + T_{x2} = 2 TOF + T_{xdelay} + (t_{off,x1} + t_{off,x2})$</li>
<li>$T_{x1} - T_{x2} = t_{x1} - t_{x2} + (t_{off,x1} - t_{off,x2})$</li>
</ul>
<p><strong>阳极时间</strong></p>
<ul>
<li>$T_a = TOF + t_{off,a}$</li>
</ul>
<p><strong>组合参数 (相对时间，消除速度依赖性)</strong></p>
<p>上述绝对时间信号不仅与探测器内部传输有关，也与束流的到达时间 ($TOF$) 有关。研究探测器特征时，必须消除束流动量弥散带来的 $TOF$ 展宽。
通过将阴极信号减去阳极信号，可以抵消 $TOF$的影响：</p>
<ul>
<li>$\Delta T_{x1} = T_{x1} - T_a = t_{x1} + (t_{off,x1} - t_{off,a})$</li>
<li>$\Delta T_{x2} = T_{x2} - T_a = t_{x2} + (t_{off,x2} - t_{off,a})$</li>
<li>$\Delta T_{x1} - \Delta T_{x2} = t_{x1} - t_{x2} + (t_{off,x1} - t_{off,x2})$</li>
</ul>
<h3 id="%E6%8E%A2%E6%B5%8B%E5%99%A8%E4%B8%A4%E7%AB%AF%E4%BF%A1%E5%8F%B7%E5%85%B3%E8%81%94">探测器两端信号关联</h3><ul>
<li>$\mathbf{T_{x1} + T_{x2}} = 2 TOF + Const.$
<strong>(包含 $TOF$，与束流速度有关，分布表现为一个较宽的峰)</strong></li>
<li>$\mathbf{\Delta T_{x1} + \Delta T_{x2}} = t_{x1} + t_{x2} + (t_{off,x1} + t_{off,x2} - 2t_{off,a}) = T_{xelay} + Const. = \mathbf{Const.}$
<strong>(与束流速度无关，分布表现为一个极窄的峰，直接反映探测器本征时间分辨率)</strong></li>
</ul>


<h3 id="Pileup-Cut">Pileup Cut</h3><ul>
<li><p>当同时有2个或以上的粒子打在 PPAC 上时，到达 delayline 的 x 的两端的时间由不同粒子的信号给出。</p>
<ul>
<li>假设 a, b 粒子同时穿过探测器，且 a 粒子靠近 $x_1$ 侧，b 粒子靠近 $x_2$ 侧，则 $T_{x_1}$ 由 a 粒子给出，$T_{x_2}$ 则由 b 粒子给出。
(TDC 只记录与 start 信号最近的第一个信号)。</li>
</ul>
</li>
<li><p>此时，$T_{x_1}+T_{x_2} &lt; T_{delay}$。即 delayline 的两端信号传输时间之和不再是常数。这个条件可作为信号堆积的标志。</p>
</li>
<li><p>从上述讨论可知，PPAC 无堆积条件：前后两个粒子到达探测器的时间间隔大于 $T_{delay}$。</p>
</li>
<li><p>一般选择加合峰（正常单粒子事件）两侧三倍 $\sigma$ 的范围作为好事件的筛选条件，剔除范围外的堆积信号：`</p>
</li>
</ul>
<p>sum gate 检查的是各路时间是否符合单事例的读出关系，可以抑制一部分 pileup，但不能识别所有重叠脉冲；窗口也会影响有效事件的保留比例。</p>

```c++
TString scut_xpile = Form("abs(dt_x2 + dt_x1 - %f) < 3 * %f", xpeak, xsigma);
TString scut_ypile = Form("abs(dt_y2 + dt_y1 - %f) < 3 * %f", ypeak, ysigma);
```

```c++
TCut cut_xpile = scut_xpile.Data() && cut_xvalid; 
TCut cut_ypile = scut_ypile.Data() && cut_yvalid; 
TCut cut_xypile = cut_xpile && cut_ypile;
```

```c++
TCanvas *c1 = new TCanvas;
tree->Draw("dt_y1-dt_y2:dt_x1-dt_x2",cut_xypile, "colz");
gPad->SetLogy(0);
gPad->SetLogz();
c1->Draw();
```
