<h1 id="Relativistic-Kinematics">Relativistic Kinematics</h1><p>This section builds the tools used to reconstruct reaction events: energy and momentum, Lorentz boosts, and four-vectors. The examples use ROOT C++. Read them in order: first construct one particle, then add particles, and finally change the reference frame. We use $c=1$; a value in GeV must never be added to one in MeV.</p>
<h3 id="Lorentz-Transformation">Lorentz Transformation</h3><p>In special relativity, the space-time coordinates of an event depend on the inertial frame in which they are measured. Consider two inertial frames: $S$, and $S'$, where $S'$ moves with velocity $v$ along the $+z$-axis relative to $S$. The Lorentz transformation relates the coordinates $(ct,x,y,z)$ and $(ct',x',y',z')$ by</p>
<p>$$
\begin{align}
x' &amp;= x, &amp; x &amp;= x' \\
y' &amp;= y, &amp; y &amp;= y' \\
z' &amp;= \gamma (z - \beta ct), &amp; z &amp;= \gamma (z' + \beta ct') \\
ct' &amp;= \gamma (ct - \beta z), &amp; ct &amp;= \gamma (ct' + \beta z')
\end{align}
$$
where</p>
<ul>
<li>$\beta = v/c$,</li>
<li>$\gamma = 1/\sqrt{1-\beta^2}$.</li>
</ul>
<p align="left">
<img alt="No description has been provided for this image" src="coordinates.png" width="200"/>
</p>
<p>A few points are especially important in physics analysis.</p>
<p>First, the space-time interval</p>
<p>$$
s^2=(ct)^2-x^2-y^2-z^2
$$</p>
<p>is invariant under Lorentz transformation. This invariance is the prototype for the invariant norm of the energy-momentum 4-vector discussed below.</p>
<p>Second, the transverse coordinates are unchanged by a boost along the $z$-axis:</p>
<p>$$
x'=x,\qquad y'=y.
$$</p>
<p>Third, the Lorentz transformation leads to two standard relativistic effects:</p>
<ul>
<li><strong>length contraction</strong> along the boost direction,</li>
<li><strong>time dilation</strong> for moving clocks.</li>
</ul>
<p>If a particle has proper lifetime $\tau$ in its rest frame, then in a frame where it moves with Lorentz factor $\gamma$, its average observed lifetime is $\gamma \tau$. This is directly relevant in detector-based reconstruction of unstable particles.</p>
<h3 id="Relativistic-Energy-and-Momentum">Relativistic Energy and Momentum</h3><p>For a particle with rest mass $m$ and velocity $v$,</p>
<p>$$
\vec{p}=\gamma m \vec{v},
\qquad
E=\gamma mc^2.
$$</p>
<p>The total energy contains both rest energy and kinetic energy:</p>
<p>$$
E = T + mc^2,
\qquad
T = (\gamma-1)mc^2.
$$</p>
<p>The energy and momentum transform under Lorentz boosts in the same way as the time and longitudinal spatial coordinates. For a boost along $z$,</p>
<p>$$
\begin{align}
p'_x &amp;= p_x, &amp; p_x &amp;= p'_x \\
p'_y &amp;= p_y, &amp; p_y &amp;= p'_y \\
p'_z &amp;= \gamma (p_z-\beta E/c), &amp; p_z &amp;= \gamma (p'_z+\beta E'/c) \\
E'/c &amp;= \gamma(E/c-\beta p_z), &amp; E/c &amp;= \gamma(E'/c+\beta p'_z)
\end{align}
$$
The most important invariant relation is</p>
<p>$$
E^2 = (pc)^2 + (mc^2)^2.
$$</p>
<p>This equation is central in all later analyses. It is often used in three equivalent ways:</p>
<p>$$
m^2c^4 = E^2 - p^2c^2,
$$</p>
<p>$$
p = \gamma \beta mc,
$$</p>
<p>and, if the kinetic energy $T$ is known instead of the total energy,</p>
<p>$$
(pc)^2 = T^2 + 2mc^2T.
$$</p>
<p>The last form is especially useful in nuclear physics, where detector calibration often yields kinetic energy more directly than total energy.</p>
<p>For a massless particle such as a photon, $m=0$, so</p>
<p>$$
E = pc.
$$</p>
<p>For highly relativistic particles with $E \gg mc^2$, the approximation $E \approx pc$ becomes accurate.</p>
<h3 id="Natural-Units">Natural Units</h3><p>In nuclear and particle physics, it is standard to use natural units with $c=1$. In this convention, energy, mass, and momentum are all expressed in units such as MeV or GeV, and many formulas simplify:</p>
<p>$$
E^2 = p^2 + m^2,
\qquad
m^2 = E^2 - p^2,
\qquad
T = E-m,
\qquad
p^2 = T^2 + 2mT.
$$</p>
<p>This convention will be used throughout the remainder of the chapter unless otherwise noted.</p>
<p>Published experimental papers still often write masses in GeV/$c^2$ and momenta in GeV/$c$ to make dimensions explicit. In analysis code, however, one usually works in natural units and omits the factors of $c$, as long as units are handled consistently.</p>
<h3 id="Example">Example</h3><p>Consider a $K_S$ meson with rest mass $m_{K_S}=0.4977$ GeV moving along the $+z$-axis with $v=0.95c$. Then</p>
<p>$$
\gamma = \frac{1}{\sqrt{1-0.95^2}} = 3.2026,
$$</p>
<p>$$
E = \gamma m_{K_S} = 3.2026 \times 0.4977 = 1.5939\ \text{GeV},
$$</p>
<p>$$
p_z = \gamma \beta m_{K_S} = 3.2026 \times 0.95 \times 0.4977 = 1.5142\ \text{GeV}.
$$</p>
<p>Since $p_x=p_y=0$,</p>
<p>$$
m = \sqrt{E^2-p^2}
   = \sqrt{(1.5939)^2-(1.5142)^2}
   = 0.4977\ \text{GeV},
$$</p>
<p>which reproduces the rest mass.</p>
<p>This is a simple but important check: in practical analysis, whenever a 4-vector is constructed, one should be able to recover the particle mass from $E^2-p^2$, up to numerical precision and detector resolution.</p>

<h2 id="Vectors-and-Lorentz-Transformations">Vectors and Lorentz Transformations</h2><p>The 4-vector formalism provides the natural language for relativistic analysis. It unifies energy and momentum into one object and makes invariant quantities explicit.</p>
<h3 id="Definition-of-4-Vectors">Definition of 4-Vectors</h3><p>A 4-vector combines one time-like component with three spatial components. Two examples are used repeatedly in relativistic mechanics:</p>
<ul>
<li><p>space-time 4-vector:
$$
X^\mu = (ct, x, y, z),
$$</p>
</li>
<li><p>energy-momentum 4-vector:
$$
P^\mu = (E/c, p_x, p_y, p_z).
$$</p>
</li>
</ul>
<p>In natural units, the energy-momentum 4-vector is usually written as</p>
<p>$$
P^\mu = (E,\vec{p}) = (E,p_x,p_y,p_z).
$$</p>
<p>For a general 4-vector $A^\mu=(A_0,\vec{A})$, the Minkowski inner product is</p>
<p>$$
A\cdot B = A_0B_0 - \vec{A}\cdot\vec{B},
$$</p>
<p>and the invariant norm is</p>
<p>$$
A^2 = A\cdot A = A_0^2 - |\vec{A}|^2.
$$</p>
<p>For the energy-momentum 4-vector,</p>
<p>$$
P^2 = E^2 - p^2 = m^2
$$</p>
<p>in natural units. This is the origin of invariant mass.</p>
<h3 id="Lorentz-Transformation-of-4-Vectors">Lorentz Transformation of 4-Vectors</h3><p>A Lorentz transformation maps one 4-vector to another in a different inertial frame while preserving the Minkowski norm. This is why the quantity $P^2=m^2$ is frame-independent.</p>
<p>In the particle rest frame,</p>
<p>$$
P^\mu = (m,0,0,0),
$$</p>
<p>so</p>
<p>$$
P^2 = m^2.
$$</p>
<p>In any boosted frame, the components $E$ and $\vec{p}$ change, but the invariant mass does not.</p>
<p>This fact is fundamental in data analysis. Peak structures in invariant-mass spectra are meaningful precisely because invariant mass is independent of the reference frame used to describe the event.</p>
<h3 id="Multi-Particle-Systems">Multi-Particle Systems</h3><p>For a system of particles, the total 4-momentum is</p>
<p>$$
P^\mu_{\text{tot}} = \sum_i P^\mu_i.
$$</p>
<p>Its invariant norm defines the invariant mass of the system:</p>
<p>$$
M_{\text{sys}}^2 = P_{\text{tot}}^2
= \left(\sum_i E_i\right)^2 - \left|\sum_i \vec{p}_i\right|^2.
$$</p>
<p>This formula is one of the most important working equations in experimental particle and nuclear physics. It is used to reconstruct short-lived resonances and composite systems from detected final-state particles.</p>
<p>For example, in</p>
<p>$$
\pi^0\rightarrow \gamma\gamma,
$$</p>
<p>the invariant mass of the two photons should peak near the $\pi^0$ mass.</p>
<h3 id="Conservation-in-Reactions">Conservation in Reactions</h3><p>For decays and scattering reactions, 4-momentum conservation is written compactly as</p>
<p>$$
P^\mu_{\text{initial}} = P^\mu_{\text{final}}.
$$</p>
<p>For example:</p>
<ul>
<li><p>decay:
$$
P \to P_1 + P_2 + \cdots
$$</p>
</li>
<li><p>scattering:
$$
P_1 + P_2 \to P_3 + P_4 + \cdots
$$</p>
</li>
</ul>
<p>Because each $P^\mu$ contains both energy and momentum, one equation automatically enforces both energy conservation and momentum conservation.</p>
<p>This compact notation will be used heavily in Sections 4.2 and 4.3, especially when defining invariant mass and missing mass.</p>

<h2 id="Physics-Vectors">Physics Vectors</h2><p>Vectors are used throughout physics analysis to describe positions, directions, momenta, and relativistic 4-momenta. In ROOT, dedicated vector classes make these quantities easy to construct, manipulate, and interpret in a way that matches standard analysis practice.</p>
<h3 id="Coordinate-Representations">Coordinate Representations</h3><p>A physical vector can be expressed in different coordinate systems depending on the problem. In collider physics, the variables $(p_T,\eta,\phi,m)$ are often convenient. In many nuclear-physics and fixed-target applications, however, the Cartesian form $(p_x,p_y,p_z,E)$ is usually the most transparent because it connects directly to momentum conservation and reaction kinematics.</p>
<p>For the present chapter, Cartesian 4-vectors are the most natural choice. They are especially convenient for writing conservation laws, constructing invariant masses, and performing explicit boosts between the laboratory and center-of-mass frames.</p>

<h2 id="Constructing-a-Four-Vector-from-Momentum-and-Mass">Constructing a Four-Vector from Momentum and Mass</h2><p>If the momentum and particle mass are known, use</p>
<div class="highlight"><pre><span></span><span class="n">SetXYZM</span><span class="p">(</span><span class="n">px</span><span class="p">,</span><span class="n">py</span><span class="p">,</span><span class="n">pz</span><span class="p">,</span><span class="n">m</span><span class="p">)</span>
</pre></div>
<p>ROOT calculates</p>
<p>$$
E=\sqrt{p_x^2+p_y^2+p_z^2+m^2}.
$$</p>
<p>Example:</p>
<p>The next two examples construct the same type of object from different measurements. <code>SetXYZM</code> is convenient when momentum is known; when kinetic energy is measured, calculate the momentum magnitude first.</p>

```c++
{
const double mpi = 0.13957; // GeV

    TLorentzVector pion;
    pion.SetXYZM(0.10, -0.05, 0.40, mpi);

    std::cout << "E = " << pion.E() << " GeV" << std::endl;
    std::cout << "M = " << pion.M() << " GeV" << std::endl;
}
```

<h2 id="Constructing-a-Four-Vector-from-Kinetic-Energy">Constructing a Four-Vector from Kinetic Energy</h2><p>If kinetic energy $T$, mass $m$, and direction $(\theta,\phi)$ are known,</p>
<p>$$
E=T+m,
$$</p>
<p>$$
p=\sqrt{T^2+2mT}.
$$</p>
<p>Example for an alpha particle:</p>
<p>Here the alpha-particle mass and kinetic energy are both in MeV. <code>SetMagThetaPhi(p,theta,phi)</code> turns the momentum magnitude and direction into Cartesian components. The final <code>M()</code> check should recover the input alpha mass.</p>

```c++
{
const double mAlpha = 3727.38; // MeV
    const double T = 5.0;          // MeV

    double theta = 40.0*TMath::DegToRad();
    double phi   = 30.0*TMath::DegToRad();

    double E = T + mAlpha;
    double p = std::sqrt(T*T + 2.0*mAlpha*T);

    TVector3 pvec;
    pvec.SetMagThetaPhi(p, theta, phi);

    TLorentzVector alpha;
    alpha.SetPxPyPzE(pvec.X(), pvec.Y(), pvec.Z(), E);

    std::cout << "M = " << alpha.M() << " MeV" << std::endl;
}
```

<h2 id="Invariant-Mass-with-TLorentzVector">Invariant Mass with <code>TLorentzVector</code></h2><p>For two particles,</p>
<div class="highlight"><pre><span></span><span class="n">TLorentzVector</span><span class="w"> </span><span class="n">pair</span><span class="w"> </span><span class="o">=</span><span class="w"> </span><span class="n">p1</span><span class="w"> </span><span class="o">+</span><span class="w"> </span><span class="n">p2</span><span class="p">;</span>
<span class="kt">double</span><span class="w"> </span><span class="n">mass</span><span class="w"> </span><span class="o">=</span><span class="w"> </span><span class="n">pair</span><span class="p">.</span><span class="n">M</span><span class="p">();</span>
</pre></div>
<p>This implements</p>
<p>$$
M^2=(E_1+E_2)^2
-
|\vec p_1+\vec p_2|^2.
$$</p>
<p>Example:</p>
<p>Adding four-vectors adds <strong>energies and momentum components</strong>, not masses. The two-pion system can therefore have a mass larger than $2m_\pi$; the difference reflects their relative motion.</p>

```c++
{
const double mpi = 0.13957; // GeV

    TLorentzVector piPlus;
    TLorentzVector piMinus;

    piPlus.SetXYZM(0.20, 0.10, 0.50, mpi);
    piMinus.SetXYZM(-0.15, -0.05, 0.30, mpi);

    TLorentzVector pair_pi = piPlus + piMinus;

    std::cout << "M(pi+ pi-) = "
              << pair_pi.M()
              << " GeV" << std::endl;
}
```

<h2 id="17.-Boosts-in-ROOT">Boosts in ROOT</h2><p>Suppose a parent particle $A$ moves in the lab frame and a daughter particle $b$ is observed either in the parent rest frame or in the lab frame.</p>
<p>We store the parent four-momentum in the lab frame as</p>
<div class="highlight"><pre><span></span><span class="n">TLorentzVector</span><span class="w"> </span><span class="n">parent_lab</span><span class="p">;</span>
</pre></div>
<p>Then</p>
<div class="highlight"><pre><span></span><span class="n">TVector3</span><span class="w"> </span><span class="n">beta</span><span class="w"> </span><span class="o">=</span><span class="w"> </span><span class="n">parent_lab</span><span class="p">.</span><span class="n">BoostVector</span><span class="p">();</span>
</pre></div>
<p>returns the velocity of the parent in the lab frame,</p>
<p>$$
\vec\beta_A = \frac{\vec p_A}{E_A}.
$$
This <code>beta</code> connects two frames:</p>
<ul>
<li>the <strong>parent rest frame</strong>, where $A$ is at rest;</li>
<li>the <strong>lab frame</strong>, where $A$ is moving.</li>
</ul>
<p>Now let <code>daughter</code> be the four-vector of a daughter particle.</p>
<p>If <code>daughter</code> is defined in the parent rest frame, then</p>
<div class="highlight"><pre><span></span><span class="n">daughter</span><span class="p">.</span><span class="n">Boost</span><span class="p">(</span><span class="n">beta</span><span class="p">);</span>
</pre></div>
<p>transforms it to the lab frame.</p>
<p>If <code>daughter</code> is given in the lab frame, then</p>
<div class="highlight"><pre><span></span><span class="n">daughter</span><span class="p">.</span><span class="n">Boost</span><span class="p">(</span><span class="o">-</span><span class="n">beta</span><span class="p">);</span>
</pre></div>
<p>transforms it back to the parent rest frame.</p>
<p>So the same <code>beta</code> is used in both directions:</p>
<ul>
<li><code>Boost(beta)</code>: parent rest frame $\rightarrow$ lab frame;</li>
<li><code>Boost(-beta)</code>: lab frame $\rightarrow$ parent rest frame.</li>
</ul>

<h2 id="18.-Example">Example: boost and inverse boost</h2><p>The parent particle has mass</p>
<p>$$
M = 1.0\ \text{GeV}
$$</p>
<p>and moves in the lab along $+z$ with</p>
<p>$$
p_z = 2.0\ \text{GeV}.
$$</p>
<p>A daughter particle is first defined in the parent rest frame. We then boost it to the lab frame, and finally boost it back to the parent rest frame.</p>
<p>ROOT's <code>Boost(beta)</code> uses the active-boost convention. Thus the plus sign here is consistent with the inverse transformation written at the start of this section. Make a copy before calling <code>Boost</code>, because it changes that four-vector in place. The example checks both invariants: the daughter mass stays fixed, and boosting back recovers all four original components.</p>

```c++
{
// Parent four-vector in the lab frame
    const double M  = 1.0;   // GeV
    const double pz = 2.0;   // GeV

    TLorentzVector parent_lab;
    parent_lab.SetPxPyPzE(0.0, 0.0, pz, std::sqrt(M*M + pz*pz));

    // Velocity of the parent in the lab
    TVector3 beta = parent_lab.BoostVector();

    // Daughter four-vector in the parent rest frame
    const double md = 0.2;   // GeV
    const double px = 0.3;   // GeV

    TLorentzVector daughter_rest;
    daughter_rest.SetPxPyPzE(px, 0.0, 0.0, std::sqrt(md*md + px*px));

    // Boost from parent rest frame to lab frame
    TLorentzVector daughter_lab = daughter_rest;
    daughter_lab.Boost(beta);

    // Boost back from lab frame to parent rest frame
    TLorentzVector daughter_back = daughter_lab;
    daughter_back.Boost(-beta);

    std::cout << "Daughter in parent rest frame:\n"
              << "  (E, px, py, pz) = (" << daughter_rest.E() << ", "
              << daughter_rest.Px() << ", " << daughter_rest.Py() << ", "
              << daughter_rest.Pz() << "); M = " << daughter_rest.M() << " GeV" << std::endl;

    std::cout << "Daughter in lab frame:\n"
              << "  (E, px, py, pz) = (" << daughter_lab.E() << ", "
              << daughter_lab.Px() << ", " << daughter_lab.Py() << ", "
              << daughter_lab.Pz() << "); M = " << daughter_lab.M() << " GeV" << std::endl;

    std::cout << "Boosted back to parent rest frame:\n"
              << "  (E, px, py, pz) = (" << daughter_back.E() << ", "
              << daughter_back.Px() << ", " << daughter_back.Py() << ", "
              << daughter_back.Pz() << "); M = " << daughter_back.M() << " GeV" << std::endl;
}
```

```c++
{
// Parent four-vector in the lab frame
    TLorentzVector parent_lab;
    parent_lab.SetPxPyPzE(0.0, 0.0, 2.0, std::sqrt(5.0));

    // Daughter four-vector measured in the lab frame
    TLorentzVector daughter_lab;
    daughter_lab.SetXYZM(0.3, 0.0, 0.8, 0.2);

    // Velocity of the parent in the lab frame
    TVector3 beta = parent_lab.BoostVector();

    // Boost daughter from lab frame to parent rest frame
    TLorentzVector daughter_rest = daughter_lab;
    daughter_rest.Boost(-beta);

    std::cout << "daughter mass = "
              << daughter_rest.M()
              << " GeV" << std::endl;

    std::cout << "theta in parent rest frame = "
              << daughter_rest.Theta()*TMath::RadToDeg()
              << " deg" << std::endl;
}
```
